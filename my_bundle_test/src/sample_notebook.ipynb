# Default notebook

This default notebook is executed using a Lakeflow job as defined in resources/sample_job.job.yml.

```python
catalog_ = dbutils.widgets.get("catalog")
schema = dbutils.widgets.get("schema")
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

spark = SparkSession.builder \
    .appName("DynamicDeltaWrite") \
    .getOrCreate()

# Dynamic input data
data = [
    (1, "John", 25, "IT"),
    (2, "David", 30, "HR"),
    (3, "Smith", 28, "Finance")
]

columns = ["id", "name", "age", "department"]

df = spark.createDataFrame(data, columns)
```

```python
table_name = "employee"
write_mode = "overwrite"

# Create database if not exists
spark.sql(f"CREATE DATABASE IF NOT EXISTS {schema}")

# Dynamic table name
full_table_name = f"{catalog_}.{schema}.{table_name}"

# Write DataFrame into Delta table
df.write \
    .format("delta") \
    .mode(write_mode) \
    .option("overwriteSchema", "true") \
    .saveAsTable(full_table_name)

print(f"Data successfully written to {full_table_name}")
```
